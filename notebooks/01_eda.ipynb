# Análisis Exploratorio de Datos

## 1. Carga y exploración inicial

En esta sección se realiza una exploración inicial del conjunto de datos de navegación de un robot móvil. Se analiza la cantidad de registros, los tipos de datos, las estadísticas descriptivas y la distribución de las clases de navegación.

## 2. Estructura del dataset

El conjunto de datos contiene 5456 registros y 25 columnas. Las primeras 24 columnas corresponden a mediciones numéricas obtenidas mediante sensores ultrasónicos, mientras que la última columna corresponde a la clase de movimiento del robot.

## 3. Variable objetivo y balance de clases

La variable objetivo corresponde a la clase de navegación del robot. Se identifican cuatro comportamientos: avanzar, realizar un giro pronunciado hacia la derecha, realizar un giro ligero hacia la derecha y realizar un giro ligero hacia la izquierda.

La distribución de las clases no es uniforme, por lo que el conjunto de datos presenta un desbalance entre las categorías.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json
import os
```

```python
df = pd.read_csv(
    "../data/raw/sensor_readings_24.data",
    header=None
)

df.head()
```

```python
df.shape
```

```python
sensor_columns = [f"US{i}" for i in range(1, 25)]

df.columns = sensor_columns + ["Movimiento"]

df.head()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="US1",
    bins=30,
    kde=True
)

plt.title("Distribución del sensor US1")
plt.xlabel("Lectura del sensor")
plt.ylabel("Frecuencia")

plt.show()
```

```python
sensores_hist = [
    "US1",
    "US2",
    "US3",
    "US4",
    "US5",
    "US6"
]

for sensor in sensores_hist:

    plt.figure(figsize=(8, 5))

    sns.histplot(
        data=df,
        x=sensor,
        bins=30,
        kde=True
    )

    plt.title(f"Distribución del sensor {sensor}")
    plt.xlabel("Lectura del sensor")
    plt.ylabel("Frecuencia")

    plt.show()
```

```python
plt.figure(figsize=(8, 4))

sns.boxplot(
    x=df["US1"]
)

plt.title("Boxplot del sensor US1")
plt.xlabel("Lectura del sensor")

plt.show()
```

```python
sensores_boxplot = [
    "US1",
    "US2",
    "US3",
    "US4"
]

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df[sensores_boxplot]
)

plt.title("Distribución de los sensores US1-US4")
plt.xlabel("Sensor")
plt.ylabel("Lectura")

plt.show()
```

```python
correlacion = df[sensor_columns].corr()

correlacion
```

```python
plt.figure(figsize=(14, 11))

sns.heatmap(
    correlacion,
    cmap="coolwarm",
    center=0
)

plt.title("Matriz de correlación entre sensores")

plt.show()
```

```python
plt.figure(figsize=(14, 11))

sns.heatmap(
    correlacion,
    cmap="coolwarm",
    center=0
)

plt.title("Matriz de correlación entre sensores")

plt.show()
```

```python
correlacion_abs = correlacion.abs()

pares_correlacion = (
    correlacion_abs
    .where(np.triu(np.ones(correlacion.shape), k=1).astype(bool))
    .stack()
    .sort_values(ascending=False)
)

pares_correlacion.head(10)
```

```python
with open(
    "../data/processed/eda_summary.json",
    "w",
    encoding="utf-8"
) as archivo:

    json.dump(
        eda_summary,
        archivo,
        indent=4,
        ensure_ascii=False
    )
```
