# Clase 6 — Comparación de modelos supervisados

## Random Forest vs SVM

En esta actividad se comparan dos modelos de aprendizaje supervisado
para clasificar los movimientos del robot a partir de las lecturas
de los sensores.

Los modelos evaluados son:

- Random Forest
- Support Vector Machine (SVM)

Las métricas utilizadas son Accuracy, F1-score y tiempo de entrenamiento.

```python
import pandas as pd

resultados = pd.DataFrame({
    "Modelo": ["Random Forest", "SVM"],
    "Accuracy": [0.9954, 0.8690],
    "F1-score": [0.9954, 0.8684],
    "Tiempo (s)": [0.5063, 0.2967]
})

resultados
```

```python
import matplotlib.pyplot as plt

resultados_grafico = resultados.set_index("Modelo")[["Accuracy", "F1-score"]]

ax = resultados_grafico.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Comparación de modelos supervisados")
plt.xlabel("Modelo")
plt.ylabel("Puntuación")
plt.ylim(0, 1.05)
plt.xticks(rotation=0)
plt.legend(title="Métrica")
plt.tight_layout()
plt.show()
```

## Conclusiones

La comparación realizada entre Random Forest y SVM muestra que Random Forest
presentó el mejor desempeño para la clasificación de los movimientos del robot.

Random Forest obtuvo un Accuracy de 0.9954 y un F1-score de 0.9954, mientras
que SVM alcanzó un Accuracy de 0.8690 y un F1-score de 0.8684.

Aunque SVM presentó un menor tiempo de entrenamiento, con 0.2967 segundos
frente a 0.5063 segundos de Random Forest, la diferencia en el rendimiento
predictivo es considerable.

Por lo tanto, para este conjunto de datos se selecciona Random Forest como
el modelo más adecuado, debido a que logra clasificar los movimientos del
robot con mayor precisión y un mejor equilibrio entre las clases.
